### 1. API와 HTTP 정리
- API는 프로그램끼리 데이터를 주고 받는 약속
- 예) 식당 주문 -> 테이블 태블릿에서 주문하면(주문 방식이 공통: API) -> 음식이 옴 (데이터 받기)
---
- HTTP : 프로그램이 서버에 요청을 보내는 규칙
- GET : 데이터 조회할 때
- POST : 데이터 전송할 때

### 2. requests 외부 라이브러리로 GET 요청
- 설치 : uv add requests
- 주소(URL)에 GET 요청을 보내면 JSON으로 답이 온다

```python
import requests # 모듈 불러오기 - 외부 모듈 -> 사전 설치

url = "https://jsonplaceholder.typicode.com/todos/1"

res = requests.get(url)     # GET 방식으로 요청 - res에 담음
status = res.status_code    # 상태코드 - 200
data = res.json()           # 실제 본문 데이터 -> json 형태로 변환

print("상태코드", status)
print("실제 데이터", data)
```

```python
# 테스트 한번 더 - params

url = "https://jsonplaceholder.typicode.com/comments"

res = requests.get(url, 
                   params={"postId" : 1})

data = res.json()
print(data)
```

```python
type(data) # json으로 받은 결과는 리스트!
```

```python
data[0]
```

```python
emails = []

for item in data:
    emails.append(item["email"])

emails
```

```python
# 실습
# 1. 위 내용에서 body 내용만 담긴 리스트 만들기
# 2. https://jsonplaceholder.typicode.com/todos 주소에서 userId가 1인 사람의 타이틀만 저장해보기
```

```python
# 1
bodys = []

for item in data:
    bodys.append(item["body"])

bodys
```

```python
# 2
url = "https://jsonplaceholder.typicode.com/todos"
res = requests.get(url,
                   params={"userId" : 1})
data = res.json()

titles = []

for item in data:
    titles.append(item["title"])

titles
```
