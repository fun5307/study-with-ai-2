### 공공데이터 API 사용해보기

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
service_key = os.getenv("SERVICE_KEY")
service_key
```

```python
import requests

url = "https://api.kcisa.kr/openapi/service/rest/convergence2019/getConver10"

headers = {
    "Accept" : "application/json"
}

params = {
    "serviceKey" : service_key,
    "numOfRows" : 100,
    "pageNo" : 1,
    # "keyword" : "축제",
    "where" : "서울"
}

response = requests.get(url,
                        params=params,
                        headers=headers
                        )
response.status_code
```

```python
# 헤더 설정 안하면이 데이터는 json이 아니라 response.json()이 안된다.!
print(response.text)
```

```python
# 헤더 설정 하니까 json 가능하다!

data = response.json()
data['response']['body']['items']['item'] # 처음 data는 딕셔너리 key가 많은 상황이라 하나씩 들어가줘야 깔끔하게 나온다!
```

- pageNo : 1
- numOfRows : 10

```python
# for 문 짜보기 - while로 해도 됨

url = "https://api.kcisa.kr/openapi/service/rest/convergence2019/getConver10"

headers = {
    "Accept" : "application/json"
}

# 데이터 담을 리스트
total_data = []

# pageNo를 1~5
for pageNo in range(1,6):
    params = {
        "serviceKey" : service_key,
        "numOfRows" : 10,
        "pageNo" : pageNo,
        # "keyword" : "야경",
        "where" : "제주"
    }

    response = requests.get(url,
                            params=params,
                            headers=headers
                            )
    print(response.status_code)

    data = response.json() # 파이썬에서 사용가능한 형태로 변환
    total_data += data['response']['body']['items']['item'] # 데이터 더해줌 - 우변 최종 형태가 리스트이기 때문에 리스트끼리 합치면 됨
```
