#### 도전 1 — 좌석 예약 시스템

```python
class Seat:
  
  def __init__(self, 좌석번호, 가격):
    # TODO: 좌석번호, 가격, 예약자 속성 만들기
    self.number = 좌석번호
    self.price = 가격
    self.name = None


  def 예약하기(self, 이름):

    if self.name == None:
      self.name = 이름
      print(f"{self.name}님 예약 완료")
    elif self.name == 이름:
      print("이미 예약하셨습니다.")
    else:
      print(f"{self.number} 좌석은 이미 {self.name}님이 예약했습니다.")
    

  def 예약_취소(self, 이름):
    self.cancel = 이름
    if self.cancel == self.name:
      print(f"{self.number} 좌석 예약이 취소되었습니다")
      self.name = None
    elif self.name == None:
      print("예약된 좌석이 아닙니다.")
    else:
      print("예약자 본인만 취소할 수 있습니다.")

  def 좌석_정보(self):
    if self.name == None:
      print(f"좌석: {self.number} / 가격: {self.price} / 예약 가능")
    else:
      print(f"좌석: {self.number} / 가격: {self.price} / 예약 불가")
      
       
A1 = Seat("A1", 12000)
A2 = Seat("A2", 12000)

A1.예약하기("민지")
A1.예약하기("준호")
A1.예약_취소("준호")
A1.예약_취소("민지")

A1.좌석_정보()
A2.좌석_정보()
```

#### 도전 2 — 캐릭터 대결 게임

```python
class Character:
    def __init__(self, 이름, 체력, 공격력):
        self.이름 = 이름
        self.체력 = 체력
        self.공격력 = 공격력
         
         
	
    def 공격하기(self, 상대):
       상대.체력 -= self.공격력
       상대.체력 = max(상대.체력, 0)  # 체력이 음수가 되지 않도록 처리

       if 상대.체력 == 0:
          print(f"{상대.이름}은 쓰러졌습니다.")

    def 상태_출력(self):
        print(f"이름: {self.이름}, 체력: {self.체력}, 공격력: {self.공격력}")
      
전사 = Character("전사", 100, 25)
마법사 = Character("마법사", 80, 30)

while 전사.체력 > 0 and 마법사.체력 > 0:
    전사.공격하기(마법사)
    마법사.상태_출력()
    
    if 마법사.체력 > 0:
        마법사.공격하기(전사)
        전사.상태_출력()
    else:
        break
```

#### 도전 3 — 은행 계좌 관리 프로그램

```python
class BankAccount:
    def __init__(self, 계좌번호, 예금주, 잔액=0):
        self.계좌번호 = 계좌번호
        self.예금주 = 예금주
        self.잔액 = 잔액

    def 입금(self, 금액):
        if 금액 > 0:
            self.잔액 += 금액
            print(f"입금 완료. 잔액: {self.잔액}원")
        else:
            print("입금 금액은 0보다 커야 합니다.")
        

    def 출금(self, 금액):
        if 금액 < 0:
            print("출금 금액은 0보다 커야 합니다.")
        elif 금액 > self.잔액:
            print("잔액이 부족합니다.")
        else:
            self.잔액 -= 금액
            print(f"출금 완료. 잔액: {self.잔액}원")

    def 이체(self, 받는계좌, 금액):
        if 금액 < 0:
            print("이체 금액은 0보다 커야 합니다.")
        elif 금액 > self.잔액:
            print("잔액이 부족합니다.")
        else:
            if not isinstance(받는계좌, BankAccount):
                print("받는 계좌가 유효하지 않습니다.")
            else:
                self.잔액 -= 금액
                받는계좌.입금(금액)
                print(f"이체 완료. 잔액: {self.잔액}원")

    def 계좌_정보(self):
        print(f"계좌번호: {self.계좌번호}, 예금주: {self.예금주}, 잔액: {self.잔액}원")
      
 
민지계좌 = BankAccount("100-01", "민지", 50000)
준호계좌 = BankAccount("100-02", "준호", 10000)
계좌목록 = [민지계좌, 준호계좌]

민지계좌.이체(준호계좌, 20000)
민지계좌.계좌_정보()
준호계좌.계좌_정보()

for 계좌 in 계좌목록:
    if 계좌.예금주 == "준호":
        계좌.계좌_정보()
```
