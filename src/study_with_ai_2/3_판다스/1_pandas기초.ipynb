### 데이터프레임(DataFrame) : 표 -> 대부분의 데이터를 다루는 형태
- 엑셀처럼 행과 열로 된 표를 다루는 가장 기본 형태
- 수천, 수만줄의 데이터도 코드 몇줄로 다루는 게 가능
- 데이터 분석 해보고 싶다면 필수!

#### 1. Series - 한 줄짜리 데이터
- 리스트와 비슷하지만, 각 값마다 인덱스가 붙음
- 표의 한 열을 나타낸다

```python
# 실습
import pandas as pd

data = [81,72,33,94] # 이거 출력하면 가로로 뜸

data_series = pd.Series(data)
data_series # 이렇게 하면 세로로 뜬다!
```

```python
en_data = [92,88,67,54]
en_series = pd.Series(en_data)

en_series
```

#### 2. 데이터프레임 만들기
- 데이터프레임은 여러 시리즈를 나란히 붙인 표
- 파이썬 딕셔너리로 만들면 편하다!

```python
# 딕셔너리 자체는 value 갯수 다 같을 필요 없지만, df로 만드려면 반드시 같아야 함!
data = {
      "이름": ["이현근", "김수연", "송수림", "장안나", "백성현", "김병주", "유종혁", "주승우"],
      "성별": ["남", "여", "여", "여", "남", "남", "남", "남"],
      "나이": [27, 27, 44, 31, 29, 35, 26, 40],
      "취미": ["코딩", "숨쉬기", "맛집탐방", "독서", "축구", "영화감상", "게임", "등산"],
      "키": [182, 161, 165, 158, 175, 170, 178, 183],
      "몸무게": [62, None, None, 50, 68, 72, 75, 80]
  }
df = pd.DataFrame(data)
df
```

#### 3. 데이터의 형태를 확인 - head, tail
- 보통 데이터의 처음과 끝을 확인해볼 필요

```python
df.head(3)
```

```python
df.tail(3)
```

#### 4. 표의 크기 - shape
- shape : 행과 열의 크기
- 괄호 없이 씀 -> 메서드가 아니라 속성

```python
df.shape    # (행,열) 순서 -> 열 갯수, 행 갯수
```

#### 5. 전체 요약 정보 - info()
- 비어있지 않은 값의 갯수를 한눈에 보기

```python
df.info()
```

#### 6. 숫자 통계 요약 - describe()
- 숫자로 된 열의 평균, 최대, 최소 기본 통계값
- 문자열 섞여있는 열은 자동으로 빠짐

```python
df.describe()
```

#### 7. 타입형태 - 각 열이 어떤 자료형인지 확인할 필요
- int64, float64, object(문자형) 보통

```python
df.dtypes
```
