- 조건 걸어서 해보기
- 특정 열과 행을 선택해서 처리해야 하는 경우가 대부분

```python
import pandas as pd
```

```python
# 상위 폴더로 가려면 ../ 치면 된다!

df = pd.read_csv("../data/3-1_students.csv")
df
```

#### 3. 행 선택 - loc(이름표로 찾기)
- loc 행의 인덱스 라벨명으로 행을 꺼냅니다
- 지금 위의 데이터에서는 라벨명이 0,1,2,...

```python
df.loc[0]
```

```python
df.loc[1:3] # 1~2번째가 아니라, '이름'으로 하므로 1~3까지 다 뜸 (슬라이싱과 다르다!!)
```

```python
# 인덱스 0~5 아닌 경우도 많음
df.index = ["A", "B", "C", "D", "E", "F"]
df
```

```python
df.loc['A']
```

```python
df.loc["B":"D"]
```

```python
df.loc["B":"D"][["국어", "수학", "영어"]]
```

```python
df.loc["B":"D", ["국어", "수학", "영어"]]   # 옆에 두지 말고 쉼표 찍고 안에 넣어도 된다!
```

#### 4. 행 선택 - iloc (순서번호로 찾기)
- 우리가 알던 인덱싱, 슬라이싱!
- 몇 번째인지 순서로 꺼냄

```python
df.iloc[0]
```

```python
df.iloc[1:3]    # 슬라이싱과 같으므로, 1~2까지만 나온다!
```

```python
df.iloc[1:3, 2:4]   # 첫 번째는 행 슬라이싱, 쉼표 찍고 열 슬라이싱도 할 수 있다!
```

#### 5. 조건으로 데이터 가져오기
- 비교식을 대괄호 안에 넣으면 조건이 참인 행만 고를 수 있다

```python
df.head()
```

```python
# 이 결과가 하나의 Series로 나온다! 즉 참/거짓을 순서대로 담은 Series
mask = df['성별'] == "남"
mask
```

```python
df[mask]
```

```python
# 바로 작성도 가능
df[df["국어"] >= 85]
```

```python
df[df["수학"] >= 80]

# 결과는 에러 뜬다! 수학 점수에 "결석" 이라는 값 있기 때문... -> 전처리의 필요성!
```

#### 6. 여러 조건 - &(그리고), |(또는)
- 조건이 여러 개면 &, | 적절히 사용
- 각 조건은 소괄호로 감사야 함!

```python
# A 반이면서 국어가 85점 이상인 학생만
mask = (df["반"] == "A") & (df["국어"] >= 85)

df[mask]
```

```python
# 연습 1- 성별이 여자면서 영어 점수가 90점 이상

mask = (df["성별"] == "여") & (df["영어"] >= 90)

df[mask]
```

```python
# 연습 2 - 성별 여자거나 국어 80점 이상

mask = (df["성별"] == "여") | (df["국어"] >= 80)
df[mask]
```
