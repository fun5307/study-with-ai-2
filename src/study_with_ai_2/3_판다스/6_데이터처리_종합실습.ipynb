### 2. 데이터 정제 및 전처리

```python
columns = ["관리기관명", "설치목적구분", "카메라대수"]
clean_df = df[columns]
clean_df.groupby("설치목적구분").count()
```

```python
clean_df.isna().sum()
```

```python
camera_total = clean_df.groupby("관리기관명")[["카메라대수"]].sum()
camera_total = camera_total.reset_index()
camera_total
```

```python
df2 = pd.read_csv("../data/population.csv")
df2
```

```python
mask = (df2["동별(3)"] == "소계") 

gu_total = df2[mask]
gu = gu_total['동별(2)']
gu_total
```

```python
mask2 = (df2.iloc[1] == "소계")
clean_gu_total = gu_total.loc[:, mask2]
clean_gu_total.columns = ["세대 총합", "인구 총합", "한국인 총합", "외국인 총합"]
clean_gu_total['지역'] = gu
population_total = clean_gu_total.copy()
population_total = population_total.reset_index(drop=True)
population_total = population_total.drop(0).reset_index(drop=True)
columns = list(population_total.columns)

columns.remove("지역")
columns.remove("세대 총합")
columns.insert(0, "지역")
population_total = population_total[columns]
population_total
```

```python
df3 = pd.read_csv("../data/crime_fixed.csv", encoding="utf-8-sig")
df3
```

```python
mask = df3["발생검거"] == "발생"
clean_df3 = df3[mask]
# clean_df3

crime_total = clean_df3.groupby("구분")[["건수"]].sum()
crime_total = crime_total.reset_index()
crime_total
```

camera_total, population_total, crime_total 모였다!

```python
camera_total["관리기관명"] = camera_total["관리기관명"].str.replace(
      r"^(서울특별시|서울시)\s*",
      "",
      regex=True
  )

camera_total["관리기관명"] = camera_total["관리기관명"].str.replace(
      r"^(서울특별시|서울시)\s*|\s*청$",
      "",
      regex=True
  )

camera_total = camera_total.rename(
      columns={"관리기관명": "지역"}
  )

camera_total
```

```python
station_to_gu = {
      "강남": "강남구",
      "강동": "강동구",
      "강북": "강북구",
      "강서": "강서구",
      "관악": "관악구",
      "광진": "광진구",
      "구로": "구로구",
      "금천": "금천구",
      "남대문": "중구",
      "노원": "노원구",
      "도봉": "도봉구",
      "동대문": "동대문구",
      "동작": "동작구",
      "마포": "마포구",
      "방배": "서초구",
      "서대문": "서대문구",
      "서부": "은평구",
      "서초": "서초구",
      "성동": "성동구",
      "성북": "성북구",
      "송파": "송파구",
      "수서": "강남구",
      "양천": "양천구",
      "영등포": "영등포구",
      "용산": "용산구",
      "은평": "은평구",
      "종로": "종로구",
      "종암": "성북구",
      "중랑": "중랑구",
      "중부": "중구",
      "혜화": "종로구"
  }
```

```python
crime_with_gu = crime_total.copy()

crime_with_gu["지역"] = crime_with_gu["구분"].map(station_to_gu)

crime_with_gu
```

```python
crime_gu_total = crime_with_gu.groupby(
      "지역",
      as_index=False
  )["건수"].sum()

crime_gu_total
```

### 3. 병합 및 집계

```python
total_df = pd.merge(
      population_total,
      camera_total,
      on="지역",
      how="left"
  )
total_df
```

```python
final_df = pd.merge(
      total_df,
      crime_gu_total,
      on="지역",
      how="left"
  )
final_df
```

### 4. 인사이트 정립

```python
analysis_df = final_df.copy()
analysis_df.dtypes
```

```python
columns = ["인구 총합", "한국인 총합", "외국인 총합"]

analysis_df[columns] = analysis_df[columns].apply(pd.to_numeric)
```

```python
analysis_df['인구 1만명당 카메라 수'] = analysis_df['카메라대수'] / analysis_df['인구 총합'] * 10000
analysis_df['인구 1만명당 범죄 건수'] = analysis_df['건수'] / analysis_df['인구 총합'] * 10000
analysis_df
```

```python
# 1. 인구 1만 명당 카메라 수 순위
camera_rank = analysis_df.sort_values(
      "인구 1만명당 카메라 수",
      ascending=False
  )[["지역", "인구 1만명당 카메라 수"]]

camera_rank
```

```python
# 2. 인구 1만 명당 범죄 건수 순위
crime_rank = analysis_df.sort_values(
      "인구 1만명당 범죄 건수",
      ascending=False
  )[["지역", "인구 1만명당 범죄 건수"]]

crime_rank
```

```python
# 3. 두 지표의 상관계수
correlation = analysis_df[
      ["인구 1만명당 카메라 수", "인구 1만명당 범죄 건수"]
  ].corr().iloc[0, 1]

correlation
```

```python
camera_mean = analysis_df["인구 1만명당 카메라 수"].mean()
crime_mean = analysis_df["인구 1만명당 범죄 건수"].mean()

# 카메라도 평균보다 많고 범죄도 평균보다 많은 구
high_camera_high_crime = analysis_df[
      (analysis_df["인구 1만명당 카메라 수"] >= camera_mean) &
      (analysis_df["인구 1만명당 범죄 건수"] >= crime_mean)
  ]

high_camera_high_crime
```

```python
# 카메라는 평균보다 적은데 범죄는 평균보다 많은 구
low_camera_high_crime = analysis_df[
      (analysis_df["인구 1만명당 카메라 수"] < camera_mean) &
      (analysis_df["인구 1만명당 범죄 건수"] >= crime_mean)
]

low_camera_high_crime
```

### 최종 결론(인사이트)
- 인구 1만명당 카메라 수와 인구 1만명당 범죄 건수의 상관관계는 그렇게 높지도 않고 낮지도 않다

    - 서울시 자치구별 분석 결과, 인구 1만 명당 카메라 수와 범죄 건수 사이의 상관계수는 약 0.555로 나타났다. 두 지표는 중간 정도의 양의 상관관계를 보였지만, 이를 카메라가 범죄 발생을 증가시킨다는 인과관계로 해석할 수는 없다.

- 강남, 용산, 종로, 중구는 카메라도 많고 범죄율도 높다
    - 인과관계 정확히 설명하기 어려움. 상권·관광지·업무지구처럼 유동 인구가 많은 지역일 가능성이 큽니다. 범죄 위험이 높거나 신고·관리 수요가 많아서 카메라도 상대적으로 많이 설치됐다고 해석할 수 있다.

- 마포, 영등포구는 카메라는 적은데 범죄율은 높다.
    - 유동 인구나 상업 활동에 비해 CCTV 설치가 부족할 가능성을 살펴볼 수 있다. 조심해야 할지도...?

#### 약간의 개선사항

1. 카메라 데이터에 설치목적 보면 방범만 있는 게 아니라 교통, 쓰레기단속, 기타 등 다양한 항목이 있던데, 특정 설치구분(방범 등)만 뽑으면 조금 더 정확한 분석이 가능할 것 같다.

2. 경찰서 기준 범죄 데이터를 행정구 기준으로 변환해 사용했으므로 결과 해석에는 한계가 있다.
