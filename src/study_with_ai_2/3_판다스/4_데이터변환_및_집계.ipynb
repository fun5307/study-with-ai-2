#### 데이터 그대로 쓰지 않고 조건을 정해서 처리하기
- 공원 면적을 통계확인

```python
df.describe()   # 수치형 데이터만
```

```python
df["공원면적"].mean()
```

```python
# 공원 면적 중 평균이나 중앙값 등 다양한 대표 지표 설정할 수 있는데, 일단은 크냐 작냐로 나누자
# 데이터 몇개만 뽑기
columns = ["공원명", "공원구분", "공원면적"]

clean_df = df[columns]
clean_df
```

```python
# 칼럼 하나 새로 생성!
clean_df['큰공원여부'] = clean_df["공원면적"] >= 50000
clean_df
```

#### 2. 면적 순으로 정렬하기
- 큰 값부터 정렬 : sort_values(, ascending = False)
- 작은 값부터 정렬 : sort_values()

```python
# clean_df.sort_values("공원면적") # 이렇게 하면 작은 값부터 나온다!
clean_df.sort_values("공원면적", ascending=False) # 이렇게 하면 큰 값부터 나온다!
```

#### 3. 특정 그룹별로 집계 - (통계 계산)
- groupby
- 공원 종류별로 묶어서, 갯수와 합계, 평균 구할 수 있음

```python
# 공원 면적이 0이거나 얼마 이하인 값들은 제거해야 할 수 있음 (아직 처리 X)

clean_df.groupby("공원구분")["공원면적"].count()    # 같은 공원구분으로 묶고, 공원 면적
```

```python
# agg 함수는 원하는 것들만 뽑을 수 있다
summary = clean_df.groupby("공원구분")["공원면적"].agg(["count", "sum", "mean", "median"])
summary
```

```python
summary.sort_values("count", ascending=False)
```

#### 인사이트
- 어린이공원, 소공원은 갯수 많지만 크기가 작다
- 근린공원은 평균 면적과 전체 공원면적 매우 크다
- 개수 1위는 어린이공원이지만, 전체 면적으로 보면 근린공원이 주된 공원이라 볼 수 있다

#### 4. 교차표 만들기 - pivot_table
- 어떤 두 칼럼(범주)를 가로, 세로로 교차시켜서 요약하는 것
- 관리기관 x 공원구분의 평균 면적을 같이 보려고 할 때

```python
table_df = clean_df.pivot_table(index="공원구분", 
                                columns="큰공원여부",
                                values="공원명",
                                aggfunc="count",
                                fill_value=0
                                )

table_df
# 결과물: 인덱스는 공원 구분, 행은 큰 공원 여부 O/X만, 안에 들어갈 내용은 공원명의 갯수(count)
```

#### 5. pivot_table <-> melt
- melt는 그 반대 역할

```python
# reset_index는 
melt_df = table_df.reset_index().melt(id_vars="공원구분",
                                      var_name="큰공원여부",
                                      value_name="공원수")
melt_df
```

```python
# lambda와 apply 이용해서 더 쉽게 정리할 수 있음
clean_df['규모'] = clean_df['공원면적'].apply(lambda x : "큰공원" if x > 30000 else "작은공원")
clean_df
```

```python
# 세분화해본다면
# 3만 이하는 작은공원, 3만~5만은 중간공원, 5만 이상은 큰공원
# 이럴 땐 elif가 아니라 else 안에 또 if 넣어야 한다

clean_df['규모'] = clean_df['공원면적'].apply(lambda x : "큰공원" if x >= 50000 else ("중간공원" if x >= 30000 else "작은공원"))
clean_df
```

```python
# 사실 def로 해서 apply하는 게 제일 편하다!

def clean_func(x):
    if x >= 50000:
        return "큰공원"
    elif x >= 30000:
        return "중간공원"
    else:
        return "작은공원"

clean_df['규모함수'] = clean_df['공원면적'].apply(clean_func)
clean_df
```

```python
# cut 기능으로 할 수도 있지만... 잘 안씀
clean_df['규모cut'] = pd.cut(clean_df['공원면적'], 
                           bins = [0,30000,50000,float('inf')],   # 구간 나누기
                           labels=["작은공원", "중간공원", "큰공원"],
                           right=False)

clean_df
```
