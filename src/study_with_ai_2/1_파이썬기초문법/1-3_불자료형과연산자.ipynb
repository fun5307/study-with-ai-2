### 연산자
- 산술 연산자: +, -, * 등 -> 값을 계산할 때 쓰는 연산자
- 비교 연산자: ==, !=, <, >, <=, >= -> 같다, 다르다, 크다, 작다 (결과가 bool로 나옴)
- 논리 연산자: and, or, not -> 그리고, 또는, 아닌 : 여러 조건을 묶을 때

```python
# 같다

print(3 == 4)
print(3 != 4)
print(3 == "3")
print(3 == int("3"))
```

```python
# 연습

score = 75              # 내 점수
passed = score >= 60    # 커트라인 점수
print(passed)           # 통과 여부
```

### 논리 연산자
- and: 둘 다 참일 때
- or: 또는, 어느 하나라도 참
- not: 아닐 때 참

```python
age = 28

print((20 <= age) and (age <= 30))
```

```python
print((20 <= age) or (age <= 30))
```

```python
print(not(age == 30))
```

### 멤버십 연산자

```python
text = "오늘 날씨가 비가 많이 오네요. 월요일입니다."

# in / not in
print("오늘" in text)
print("태양" not in text)
```
