### 1. 함수
- 같은 일을 여러번 할때마다 코드를 복사해서 사용해야 할 경우
- 함수로 만들어서 재사용
- 코드를 작성하다보면 같은 일을 여러 곳에서 사용하게 됩니다
- 예)
    - 이름을 받아서 인삿말 하기 -> 다른 작업에서도 쓰일 수 가 있다.
    - 키오스크에서 이름에 맞춰서 누구누구님 안녕하세요하는 기능
    - 웹사이트에서 위와 같은 인사하는 기능


```python
def greeting_n(name, number):   # 매개변수 : name, number 를 매개변수라고, 
    for i in range(number):
        print(f"{name}님 안녕하세요")

greeting_n("김수연", 5)
```

```python
# 함수 만들기 연습하기
# 1. BMI 계산 : 체중 / (키의 제곱) 구하는 함수 만들기

def BMI(weight, height):
    bmi = weight / ((height/100)**2)
    print(bmi)

BMI(82,177)
```

```python
# 2. 점수를 입력받아서 합격/불합격 출력하는 함수 만들기

def Pass(score):
    if score >= 70:
        print("합격입니다!")
    else:
        print("불합격입니다.")

Pass(100)
```

```python
# 리턴 없는 경우
def my_sum(a,b):
    total = a + b
    print(total)

my_sum(10,20)
```

```python
# 리턴 없는 함수는 결과 할당할 수 없다!
result1 = my_sum(10,20)
print(result1)
```

```python
# 리턴 있는 경우
def my_sum_2nd(a,b):
    return a+b

my_sum_2nd(10,20)
```

```python
# 리턴 있는 함수는 결과 할당 가능!
result2 = my_sum_2nd(10,20)
print(result2)
```

#### 지역변수와 전역변수

```python
count = 20

def cal(a,b):
    return count + a + b

cal(10,20)
```

```python
# 전역 변수를 함수 안에서 바꾸려 하면 오류 뜬다!
count = 20

def cal(a,b):
    count += 50
    return count + a + b

cal(10,20)
```

```python
# 전역변수 함수 안에서 바꾸고 싶을 땐 global 사용한다! (잘 안쓰고 보통 class에서 관리함)

count = 20

def cal(a,b):
    global count
    count += 50
    return count + a + b

print(cal(10,20))
print(count)
```

```python
# 주민등록번호 넣어서 남자,여자 판별하는 함수 만들기

def ID_check(id_number):
    if int(id_number[7]) % 2 == 0:
        gender = "여자"
    else:
        gender = "남자"

    return gender

ID_check("011117-3xxxxxx")
```

```python
# 기본값 지정할 변수는 가장 마지막에 놔야 한다!

def greeting2(location, age, name = "김민수"):
    return f"{name}님 안녕하세요. {location}에 사시죠? 나이가 {age}"

# 인자 넣을 때 변수 이름 앞에 두면, 순서 뒤바뀌어도 된다!
greeting2(name="주승우", age=19, location="수원")
```

```python
# args: 매개변수의 갯수가 정해져있지 않은 경우, 앞에 *을 붙인다!

def members(*args):
    members_list = list(args)
    return members_list
```

```python
# 사람 몇 명 넣든 계속 담을 수 있다!
members("수림", "수연", "종혁", "승우", "성현")
```

```python
# *args는 인수들을 튜플 형태로 받아서 정리하는데, 
# **kwargs는 함수에 전달된 인자들을 딕셔너리 형태로 받아서 처리한다!

def user_info(**kwargs):
    print(kwargs)
```

```python
user_info(name = "주승우", age=26, location="수원", hobby=["음악","배드민턴"])
```

```python
# 이때 예를 들어, 이름과 나이는 필수요소라고 하면, 혼합할 수 있다!

def user_info_2(name, age, **kwargs): # 이름과 나이는 필수이되 나머지는 막 넣을 수 있다
    print(name, age, kwargs)
```

```python
# 사용할 때, 앞의 2개는 고정이므로 꼭 '변수 이름 =' 안 붙여도 된다

user_info_2("주승우", 26, location="수원", hobby=["음악","배드민턴"])
```

```python
# 연습 - 자기소개 문장 만들기
# name, age는 필수 입력하고 추가 입력은 kwargs로 하고 자유롭게 취미, 사는 곳 등등 출력

def introduce(name, age, **kwargs):
    print(f"이름은 {name}이고 나이는 {age}살입니다. 그 외에는... {kwargs}")

```

```python
introduce("주승우", 26, location="수원", 취미=("배드민턴","음악 듣기"), 아무숫자=3)
```

```python
# 다른 방법 - key와 value 각각 반환하기

def introduce_2(name, age, **kwargs):
    print("이름은", name)
    print("나이는", age)

    for key, value in kwargs.items():
        print(key,":", value)
```

```python
introduce_2("주승우", 26, 취미="음악", 전화번호="010-2989-5307")
```

```python
# 실습문제 1: 점수를 몇 개 전달하든 평균을 반환하는 평균_구하기() 함수를 완성하기

def 평균_구하기(*점수들):
    결과 = sum(점수들)/len(점수들)
    return 결과

결과 = 평균_구하기(80, 90, 100, 70)
print(결과)
```
