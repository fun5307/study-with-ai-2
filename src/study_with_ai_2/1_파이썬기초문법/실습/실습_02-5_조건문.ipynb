# 실습 02-5 · 조건문

강의(02-5)의 if·elif·else를 직접 풀어봅니다. (반복문은 아직 안 배웠으니 쓰지 않습니다.)

각 문제의 **빈칸을 채워** 실행하고, 예상 결과가 나오면 성공입니다. 막히면 아래 「정답 예시」를 참고하세요.

## 문제 1

`나이 = 20` 일 때, 19 이상이면 `성인`, 아니면 `미성년자`를 출력하세요.

```python
# 여기에 작성하세요
# 힌트: if 나이 >= 19: ... else: ...

age = 20

if age >= 19:
    print("성인")
else:
    print("미성년자")
```

## 문제 2

`점수 = 82` 를 학점으로: 90이상 A, 80이상 B, 70이상 C, 나머지 F 를 출력하세요.

```python
# 여기에 작성하세요
# 힌트: if/elif/elif/else

score = 82

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'F'

print(grade)
```

## 문제 3

`기온 = 28` 이 20 이상 30 미만이면 `쾌적`, 아니면 `주의`를 출력하세요.

```python
# 여기에 작성하세요
# 힌트: 논리 and 로 범위 조건

temp = 28

if temp >= 20 and temp < 30:
    print("쾌적")
else:
    print("주의")
```

## 문제 4

점수 딕셔너리 `{"민수":90,"지현":55}`에서 `"민수"` 점수가 60 이상이면 `합격`, 아니면 `불합격`을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: 딕셔너리[키]로 값을 꺼내 if 로 판정

scores = {"민수":90, "지현":55}

if scores['민수'] >= 60:
    print("합격")
else:
    print("불합격")
```

## 문제 5

`돈 = 3000`, `가격 = 4500` 일 때, 돈이 가격 이상이면 `구매 가능`, 아니면 `부족: 남은 금액`을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: if 돈 >= 가격 / else 에서 가격-돈

money = 3000
price = 4500

if money >= price:
    print("구매 가능")
else:
    print("부족:", price - money)
```
