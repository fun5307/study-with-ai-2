## 문제 1

이름 `유나`, 나이 `20`을 변수에 담고, f-문자열로 `유나님은 20세입니다`를 출력하세요.

```python
# 여기에 작성하세요
# 힌트: name, age 변수에 담고 f"{name}님은 {age}세입니다"

name = "유나"
age = 20

print(f"{name}님은 {age}세입니다")
```

## 문제 2

17을 5로 나눈 **몫**과 **나머지**, 그리고 2의 **10제곱**을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: 몫 //, 나머지 %, 거듭제곱 **

a = 17
b = 5

print("몫:", a//b)
print("나머지:", a%b)
print(2**10)
```

## 문제 3

문자열 `"Python"`에서 **첫 글자**와 **마지막 글자**, 그리고 **앞 3글자**를 출력하세요.

```python
# 여기에 작성하세요
# 힌트: 인덱싱 [0], [-1] / 슬라이싱 [:3]
text = "Python"

first = text[0]
last = text[-1]
first_three = text[:3]

print(first, last, first_three)
```

## 문제 4

문자열 `"hello world"`를 **모두 대문자로** 바꾸고, 공백을 기준으로 **단어 리스트로** 나눠 출력하세요.

```python
# 여기에 작성하세요
# 힌트: upper() / split()

text_2 = "hello world"

print(text_2.upper())
print(text_2.split())
```

## 문제 5

문자열 `"25"`를 **정수로 바꿔** 5를 더한 값을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: int() 로 형변환 후 덧셈

number = "25"

answer = print(int(number)+5)
```
