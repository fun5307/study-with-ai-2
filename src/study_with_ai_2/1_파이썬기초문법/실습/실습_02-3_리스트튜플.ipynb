각 문제의 **빈칸을 채워** 실행하고, 예상 결과가 나오면 성공입니다. 막히면 아래 「정답 예시」를 참고하세요.

## 문제 1

동물 리스트 `["사자","코끼리","펭귄","기린"]`에서 **두 번째**와 **마지막** 동물을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: 인덱스 0부터. 마지막은 [-1]
animals = ["사자", "코끼리", "펭귄", "기린"]

print(animals[1])
print(animals[-1])
```

## 문제 2

수 리스트 `[10,20,30,40,50]`에서 **가운데 세 개**(20,30,40)를 슬라이싱으로 출력하세요.

```python
# 여기에 작성하세요
# 힌트: [1:4]

nums = [10,20,30,40,50]
print(nums[1:4])
```

## 문제 3

가격 리스트 `[1200,3400,900]`에 **2500을 추가**하고, **개수·합계·평균**을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: append 후 len·sum

prices = [1200,3400,900]
prices.append(2500)

print("개수:", len(prices))
print("합계:", sum(prices))
print("평균:", sum(prices)/len(prices))
```

## 문제 4

점수 리스트 `[88,92,76,100,64]`의 **최고점·최저점**을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: max() / min()

scores = [88,92,76,100,64]
print("최고점:", max(scores))
print("최저점:", min(scores))
```

## 문제 5

튜플 `(2026, 3, 15)`을 `연`,`월`,`일` 세 변수에 나눠 담아 `2026년 3월 15일`을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: 언패킹: 연, 월, 일 = 날짜

date = (2026, 3, 15)
year, month, day = date
print(f"{year}년 {month}월 {day}일")
```
