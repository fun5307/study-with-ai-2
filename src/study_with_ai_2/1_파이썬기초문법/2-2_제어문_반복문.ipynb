### 반복문
- 같은 일을 반복할 수 있는 것이 프로그램의 특징 -> 이것을 쉽게 할 수 있도록 하는 구조!

#### 1. While
- while 조건: 실행문
- 조건이 참일 경우 계속 실행되는 코드

```python
# 1~10까지 숫자 출력

count = 1
while count <= 10:
    print(count)
    count += 1
```

```python
# 연습: 10에서 시작해서 1씩 줄여서 0보다 클동안 출력해보기

count = 10

while count > 0:
    print(count)
    count -= 1
```

```python
# break - 루프 빠져나오기
count = 0

while True:
    count += 1
    print(count)

    if count ==5:
        break
```

```python
# continue - 짝수 출력 안하기

count = 0

while count < 10:
    count += 1
    if count % 2 == 0:
        continue
    print(count)
```

#### for 문

```python
num_list = [1,2,3,4,5]

for i in num_list:
    print(i**2)
```

```python
people = ["성현", "수림", "현근", "수연", "병주"]

for name in people:
    print(f"{name}님 안녕하세요!")
```

```python
# 딕셔너리 처리

data = {"성현" : 1, "수림" : 2, "현근" : 3}

for i in data:
    print(i)

for name, number in data.items():
    print(name, number)

# 딕셔너리에서 뽑아내면 key만 나온다! keys나 values나 items도 사용 가능
```

```python
# 스트링이나 튜플도 시퀀스라서, in 뒤에 사용 가능하다!

for i in "안녕하세요":
    print(i)

for i in 'a','b','c':
    print(i)
```

```python
# 뽑을 때 인덱스 넘버도 같이 출력하기!

data = ["승우", "종혁", "수연", "병주", "현근"]

# 1. range 이용
for i in range(len(data)):
    print(i, data[i])

# 2. enumerate 이용 - data 넣으면 튜플 형태로 결과 냄
for idx, name in enumerate(data):
    print(idx, name)
```

### 3. List Comprehension
- 리스트를 만들 때 for 문으로 정리한 데이터가 필요할 수 있음
- 이때 짧은 한 줄로 표현하고 싶을 때

```python
# 1부터 5까지의 제곱을 리스트로 만들기:
# 1,4,9,16,25

a = []
for i in range(1,6):
    a.append(i**2)

print(a)

squared = [num**2 for num in range(1,6)]
squared
```

```python
# 1부터 10까지 짝수의 제곱을 리스트로:

squared_even = [num**2 for num in range(1,11) if num%2==0]
squared_even
```
