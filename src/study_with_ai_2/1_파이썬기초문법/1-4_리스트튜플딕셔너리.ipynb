### 리스트

- 여러 데이터를 담을 바구니라고 생각
- 순서를 가진 자료형!
- 추가, 삭제, 변경 등이 가능하다

```python
# 간단한 리스트 만들기
fruits = ["사과", "바나나", "포도", "복숭아"]
fruits
```

```python
print("첫번째 과일:", fruits[0])
print(fruits[:2])
print(len(fruits))
```

```python
# 실습: 간단한 리스트 만들기

numbers = [1, 2, 3, 4, 5]
numbers + [7, 9, 11]    #하지만 numbers 자체가 바뀐 건 아님
```

```python
# 가운데에 넣을 땐 insert 사용 (넣을 index, 넣을 것)
numbers.insert(2, 10)
numbers
```

```python
numbers.append(20) # 하나씩 요소들 넣을 때
numbers
```

```python
# 리스트 한번에 합칠 땐 extend!
numbers.extend([30, 40, 50])
numbers
```

```python
# 숫자로 된 리스트 쉽게 만드는 법
nums_1 = list(range(5))
nums_2 = list(range(4,11))
nums_3 = list(range(1,10,2))

print(nums_1, nums_2, nums_3)
```

```python
# 리스트 수정
nums = list(range(5))
print(nums)
nums[0] = 20
nums[1:3] = [30, 40, 50, 60]    # 이런 식으로, 왼쪽 길이와 오른쪽 길이가 달라도 할당에 문제 없다!
print(nums)
```

```python
## 리스트 삭제
nums = list(range(10,60,10))
print(nums)

# 지우기 1 - index number로 지우기
del nums[3]
print(nums)

# 지우기 2 - 일치하는 것 지우기
nums.remove(30)
print(nums)

# 지우기 3 - 빈 리스트로 할당하기
nums[:2]=[]
print(nums)
```

```python
#만약 remove에서 같은 숫자 여러개면?
nums = [1,2,3,10,3,4]
nums.remove(3)
print(nums)

# 결과: 가장 앞의 것만 지워진다!
```

```python
# 지우기 4 - pop으로 지우기 (제일 많이 씀)

item = nums.pop()
print(nums)
print(item) # 꺼낸 것 할당 가능
# 가장 마지막 것을 꺼낸다!
```

### 딕셔너리

- 딕셔너리는 key와 value로 구성되어 있다.
- key : value  형태로 연결되어 있어 key로 그 value를 가져올 수 있다.
- 딕셔너리가 필요한 이유?
    - 예) 리스트는 순서 번호로 값을 찾음 -> 점수[0] -> 누군가의 점수 궁금할 때 -> 누군가의 index 번호를 알아야 하는 번거로움
- 이름표가 붙은 값들이라고 생각하면 된다
- 원소 순서 의미 없으므로, 인덱싱과 슬라이싱 불가

- {}에 key : value로 나타낸다.
- 딕셔너리[key]로 그 값을 가져올 수 있다

```python
# 딕셔너리 만드는 3가지 방법

dict1 = {'이름':"주승우", '나이':26, '지역':"수원"}
members = (('주승우', 100), ('홍길동', 85), ('김민수', 95))
dict2 = dict(members)
dict3 = {}
dict3['김재원'] = 100
```

```python
ai_family_dict = {"기현" : "01", 
                  "수림" : "78",
                  "현근" : "00",
                  "병준" : ['게임', '개발'],
                  "수연" : {"취미":"조승우덕질", "출생년도":"00"}}
```

```python
ai_family_dict["수연"]["취미"]
```

```python
# 딕셔너리 요소 추가
ai_family_dict['종혁'] = {"취미":["게임", "보드게임"], "출생년도":"89"}
ai_family_dict
```

```python
del ai_family_dict['기현']
ai_family_dict
```

```python
print(ai_family_dict.keys())
print(ai_family_dict.values())
print(ai_family_dict.items()) # 튜플 형태로 모인 (key:value) 값들을 반환
```

```python
print(ai_family_dict.get("수림"))
print(ai_family_dict.get("수림", "데이터 없음")) # 있을 때는 그 값을 가져오고, 없을 때는 기본값(기본값 없으면 출력 X)
print(ai_family_dict.get("승우", "데이터 없음"))
```
