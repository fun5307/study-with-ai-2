### 조건문 

- if / if ~ else / if ~ elif ~ else

#### 조건문이 중요한 이유
- 지금까지 코드는 그냥 위에서 아래로 실행
- 특정 상황에 따라 실행여부를 결정할 수 있도록 해야 할 때 조건문이 필요!
- 예)
    - 나이가 19살 이상->성인, 아니면->미성년자 출력
    - 점수가 90점 이상->A, 80점 이상->B
    - 물건 재고 있을 경우->주문 가능, 없을 경우->품절로 안내
-다양한 갈림길에서 실행 여부 정해주는 것이 if 조건문

#### 1. if - 조건

```python
start_time = 9

# 만약 출근 시간 9시 넘어가면->지각
if start_time > 9:
    print("지각입니다.")
```

```python
item = 5000
bank_account = 30000

if item < bank_account:
    print("구매 가능")
```

#### if ~ else
- 조건이 맞을 때 실행 ~ 아닐 때 실행

```python
age=26

if age > 19:
    print("성인입니다.")

else:
    print("미성년자입니다.")
```

```python
# 연습: 짝수 홀수 출력

number = 8

if number % 2 == 0:
    print("짝수입니다.")
else:
    print("홀수입니다.")
```

#### if ~ elif ~ else
- 갈림길(조건)이 여러 개일 때

```python
score = 82

if score > 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'F'

print("학점은 :", grade)
```

```python
#1. 온도에 따라서 옷차림 추천 -> 28도 이상 반팔, 15 이상 긴팔
temp = 25

if temp >= 28:
    print("반팔 추천합니다.")
elif temp >= 15:
    print("긴팔 추천합니다.")
else:
    print("패딩?")

#2. 할인 계산 -> 30만 이상 구매시 10프로 할인, 20만원 이상 구매시 5프로 할인, 10만 이상일 땐 5천원 할인
price = 130000

if price >= 300000:
    price_sale = price*0.9
elif price >= 200000:
    price_sale = price*0.95
elif price >= 100000:
    price_sale = price - 5000
else:
    price_sale = price

print(price_sale)
```

#### 4. 중첩 if
- 조건문 안의 조건문

```python
# 나이 20 이상이고 회원 가능할 때 입장 가능

age = 19
is_member = False
if age >= 20:
    if is_member == True:
        print("입장 가능")

    else:
        print("회원만 입장 가능")

else:
    print("성인만 입장 가능")

```

```python
# 영화관 심야관람

age = 10
with_adult = True

if age >= 15:
    print("관람이 가능합니다.")

else:
    if with_adult == True:
        print("보호자와 관람 가능합니다.")
    else:
        print("관람이 불가능합니다.")
```

```python
# 도서관 대출

time = 3
is_kid = True

if is_kid == True:
    if time > 7:
        print(f"{time-7}일 연체됐어요.")
    else:
        print(f"반납까지 {7-time}일 남았어요.")

else:
    if time > 14:
        print(f"{time-14}일 연체됐습니다.")

    else:
        print(f"반납까지 {14-time}일 남았습니다.")
```
