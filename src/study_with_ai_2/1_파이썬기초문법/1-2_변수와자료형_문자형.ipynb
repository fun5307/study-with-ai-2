```python
### 1단계: 첫 글자와 마지막 글자 꺼내기

product_name = "파이썬노트북"

# 첫 글자를 first_letter에 저장하세요.
first_letter = product_name[0]

# 마지막 글자를 last_letter에 저장하세요.
last_letter =  product_name[-1]

print(first_letter)
print(last_letter)

# 확인 기준: 파, 북이 차례로 출력되면 성공입니다.
```

```python
### 2단계: 주문번호에서 정보 꺼내기

# 주문번호의 규칙을 이렇게 정했다고 가정해 봅시다.

  #2026-SEOUL-042 (연도 - 지역 - 주문순서)

order_code = "2026-SEOUL-042"

# 연도 "2026"을 꺼내세요.
year = order_code[:4]

# 지역 "SEOUL"을 꺼내세요.
city = order_code[5:10]

# 주문순서 "042"를 꺼내세요.
order_number = order_code[-3:]

print(year)
print(city)
print(order_number)

#힌트: -의 위치도 문자열의 한 칸을 차지합니다. 어려우면 먼저 print(order_code[0]), print(order_code[1])처럼 한 글자씩 확인해 보세요.
```

```python
### 3단계: 뒤에서부터 꺼내기

file_name = "python_study_note.txt"

# 파일 확장자 txt를 뒤에서부터 슬라이싱해 보세요.
extension = file_name[-3:]

print(extension)

#힌트: 끝부분 전체를 가져올 때는 [-숫자:] 형태를 사용합니다.
```

```python
#문자열 포매팅 - 문자열 틀을 고정하고 바뀔 부분만 바꿀 때 포매팅 - f-string

f_greeting = f"{name}님 안녕하세요!"
print(f_greeting)
```

```python
name = "주승우"
year = 2026
f_greeting = f"{name}님 안녕하세요! {year}년입니다"
print(f_greeting)
```

```python
greeting_1 = "{name}님 안녕하세요! {year}년입니다.".format(name=name, year=year)
print(greeting_1)
```

```python
greeting_2 = "%s님 안녕하세요! %d년입니다." % (name,year)
print(greeting_2)
```

### 문자열 메서드
- 문자열에는 자주 쓰는 도구가 많이 있습니다.
- 메서드 : .을 찍어서 사용합니다
- 원본은 바뀌지 않고 새 문자열을 돌려주는
- strip, upper, lower, replace, split, len 등등

```python
#표시 형식 지정

#소수점 자릿수 지정
pi = 3.14159265
print(f"원주율은 {pi:.2f}입니다.")
```

```python
# 천 단위 구분 쉼표 지정

population = 100000000
print(f" 인구: {population:,}명")
```

```python
raw_text = "  I love you      "
print(raw_text)
```

```python
print(raw_text.strip())
```

```python
raw_strip_left = raw_text.lstrip()
print(raw_strip_left)
```

```python
raw_strip_right = raw_text.rstrip()
print(raw_strip_right)
```

```python
# 문자열 나누기

text = "2026-08-31"
text_blank = "2026 08 31"
text_slash = "2026/08/31"

print("원본: ", text.split("-"))
print("공백: ", text_blank.split())
print("슬래시: ", text_slash.split("/"))
```

```python
# 여러 변수 한번에 할당도 가능!
student_info = "주승우/26/수원"
name, age, region = student_info.split("/")
print(name, age, region)
```

```python
# / 대신 다른 걸로 변경도 가능

student_info.replace("/", "-")
```

```python
# 중요 - 하지만 원본은 바뀌지 않는다!
print(student_info)
```

```python
# 문자열 갯수 구하기

slash_count = student_info.count("/")
print(slash_count)
```

```python
# 문자열 길이 구하기

length = len(student_info)
print(length)
```

```python
# 시작, 끝 단어 확인

student_info.endswith("수원")
```

```python
student_info.startswith("홍길동")
```
