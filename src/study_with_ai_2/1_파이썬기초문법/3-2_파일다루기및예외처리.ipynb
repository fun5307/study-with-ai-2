### 파일 다루기
- 우리가 다루는 데이터는 코드 상에 있지많은 않다
- 실제 데이터는 외부 별도 파일에 들어있다.
- 이 외부 파일을 읽어오고, 수정하는 방법을 익혀야 한다

- 지금까지 저장했던 리스트와 딕셔너리는 프로그램이 끝나면 사라짐
1. csv 파일 : 엑셀에서 다루기 쉬운 형태
2. json 파일 : 웹에서 가져오는 데이터 형태
3. txt 파일 : 로그 데이터 등등 일반적인 텍스트 형태

#### 파일 만들기

```python
# 파일 만들기 -1

names = ['유재석', '박명수', '노홍철']  # 들어갈 데이터
# file = open('names.txt', 'w')   
file = open('names.txt', 'w', encoding="utf-8") # 이렇게 해야 우리가 읽을 수 있음

for item in names:
    file.write(item + '\n')
file.close()
```

```python
# 파일 만들기 -2
# with 구문 끝나면 알아서 닫힘 -> 더 편리하고 자주 사용!

names = ['유재석', '박명수', '노홍철']  # 들어갈 데이터

with open("text.txt", 'w', encoding="utf-8") as file: # 모드: w(생성), a(추가), r(읽을 때)
    for item in names:
        file.write(item + '\n')
```

#### 파일 읽어보기

```python
# 파일 읽어보기
# 1. 한 번에 읽기 - read
with open("text.txt", "r", encoding="utf-8") as file:
    text = file.read()  # 전체 텍스트파일 읽기
    print(text)
```

```python
# 2. 한줄씩 읽기 - readline : 반복문

with open("text.txt", "r", encoding="utf-8") as file:
    while True:
        # text = file.readline()  
        text = file.readline().strip() # 전체 텍스트파일 읽을 때, 뒤의 엔터 없애기

        # 종료 조건 지정해줘야 무한반복 피할 수 있음
        if text == "":
            break
        print(text)
    
```

```python
# 3. 한줄씩 읽어서 리스트로 읽기 - readlines

with open("text.txt", "r", encoding="utf-8") as file:
    text = file.readlines()  # 전체 텍스트파일 읽기
    print(text)
```

#### 파일 내용 추가하기

```python
# 파일에 내용 추가하기 : 모드를 a로 바꿔줘야

with open("text.txt", "a", encoding="utf-8") as file:
    file.write("하하")
```

### 예외처리
- 파일이 없는 경우
- 숫자가 아닌 값을 계산하라고 하는 경우
- 0으로 나누는 경우 등등
- 다양한 에러 상황을 처리할 수 있는 문법
    - try/except 사용법
    - else/finally 사용법
    - raise -> 일부러 에러를 띄우는 경우

```python
# 사용해보기
# try : 에러가 뜰 수도 있는 경우 코드
# except : 에러 상황에 처리하는 코드

try: 
    result = 10/0
except Exception as error:  # 이런 식으로 에러 코드 뭔지 받아올 수 있다!
    print("에러입니다", error)

try: 
    result = "123" + 456
except Exception as error:
    print("에러입니다", error)
    result = int("123") + 456
    print("제대로 된 결과:", result)
```
