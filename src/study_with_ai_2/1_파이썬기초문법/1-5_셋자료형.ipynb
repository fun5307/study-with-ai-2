### 셋자료형
- 중복 제거용으로 주로 쓰임
- {}로 만듦 : 딕셔너리와 구분 필요

```python
a=[1,2,2,2,3,3,3,3,3,5]
data3 = set(a)
data3
```

```python
# 합집합 - 중복 제거한 전체 값 -> 기호는 | 혹은 union 메서드 사용

data1 = {1,2,3}
data2 = {3,4,5}

print(data1 | data2)
print(data1.union(data2))
```

```python
# 교집합 - 공통 찾기
print(data1 & data2)
print(data1.intersection(data2))
```

```python
# 차집합

print(data1 - data2)
print(data1.difference(data2))
print(data2 - data1)
```
