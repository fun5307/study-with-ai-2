## 문제 5. 학점 매기기 · ⭐⭐ 기본

```python
점수 = 82

# TODO: if / elif / else 로 학점 출력

if 점수 >= 90:
    print('A')
elif 점수 >= 80:
    print('B')
elif 점수 >= 70:
    print('C')
elif 점수 >= 60:
    print('D')
else:
    print('F')

```

## 문제 9. 나이별 입장료 · ⭐⭐⭐ 응용

`나이`에 따라: **8세 미만 무료(0), 19세 미만 5000, 65세 미만 10000, 그 이상 무료(0)**. `요금`을 정해 출력하세요.

```python
나이 = 15

# TODO: 구간별로 요금 정하기
요금 = None

if 나이 >= 65:
    요금 = 0
elif 나이 >= 19:
    요금 = 10000
elif 나이 >= 8:
    요금 = 5000
else:
    요금 = 0

print("입장료:", 요금, "원")
# 예상: 입장료: 5000 원
```

## 문제 10. 배송비 · ⭐⭐⭐ 응용

`회원`이거나(`or`) `금액`이 50000 이상이면 배송비 `무료(0)`, 아니면 `3000`. `배송비`를 정해 출력하세요.

```python
금액 = 30000
회원 = True

# TODO: 회원 or 금액 50000 이상 -> 무료
배송비 = None

if 회원 == True or 금액 >= 50000:
    배송비 = 0
else:
    배송비 = 3000

print("배송비:", 배송비, "원")
# 예상: 배송비: 0 원
```

## 문제 11. 가위바위보 한 판 · ⭐⭐⭐ 응용

`나`와 `컴퓨터`의 손을 비교해 `이김`/`짐`/`비김`을 출력하세요.
(이기는 경우: 가위>보, 바위>가위, 보>바위)

```python
나 = "바위"
컴퓨터 = "가위"

# TODO: 같으면 비김, 이기는 3가지면 이김, 나머지 짐

if 나 == 컴퓨터:
    print("비김")

elif 나 == "가위":
    if 컴퓨터 == "바위":
        print("짐")
    elif 컴퓨터 == "보":
        print("이김")

elif 나 == "바위":
    if 컴퓨터 == "가위":
        print("이김")
    elif 컴퓨터 == "보":
        print("짐")

elif 나 == "보":
    if 컴퓨터 == "가위":
        print("짐")
    elif 컴퓨터 == "바위":
        print("이김")
```

## 도전 1. 미니 계산기 · ⭐⭐⭐⭐ 도전

`연산`이 `"+"`,`"-"`,`"*"`,`"/"` 중 무엇이냐에 따라 `a`,`b`를 계산해 출력하세요.

```python
a = 12
b = 4
연산 = "*"

# TODO: 연산 문자에 따라 if/elif 로 계산
결과 = None

if 연산 == "+":
    결과 = a+b
elif 연산 == "-":
    결과 = a-b
elif 연산 == "*":
    결과 = a*b
else:
    결과 = a/b

print("결과:", 결과)
# 예상: 결과: 48
```

## 도전 2. 놀이기구 탑승 · ⭐⭐⭐⭐ 도전

`키`가 140 이상 **그리고** `나이`가 10 이상이면 `탑승 가능`.
아니면 이유를 안내: 키가 작으면 `키가 작아요`, 그 외엔 `나이가 어려요`.

```python
키 = 135
나이 = 12

# TODO: and 로 둘 다 만족하면 탑승 가능, 아니면 이유 안내

if 키 >= 140 and 나이 >= 10:
    print("탑승 가능")
elif 키 < 140:
    print("키가 작아요")
else:
    print("나이가 어려요")
```
