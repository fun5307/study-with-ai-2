### 클래스
- 공통된 속성과 기능을 한번만 정의하고 이를 바탕으로 객체를 효율적으로 관리하기 위한 개념
- 설계도 개념이라고 생각

```python
# 간단한 클래스 만들기 - 메서드는 클래스에서 작동하는 함수!

class Human:
    #  속성 정의
    name = ""
    age = 0

    # 메서드 정의
    def introduce(self):
        print(f"이름은 {self.name}입니다")
        print(f"나이는 {self.age}입니다")
```

```python
minsu = Human()
minsu
```

```python
minsu.name = "김민수"
minsu.age = 25
```

```python
minsu.__dict__
```

```python
minsu.introduce()
```

```python
# 클래스 생성자 - 초기값 입력(원래는 init이 있어야 함)

class Book:
    # 초기에 반드시 있어야 하는 항목이 있을 경우
    def __init__(self, title, author):
        self.title = title   # 입력받은 title을 self.title의 속성으로 저장
        self.author = author 

    # 메서드 추가 - 간단한 책 소개 기능

    def show_info(self):
        print(f"""
이 책의 제목은 {self.title}이고,
저자는 {self.author}입니다.
많이 애용해주세요
""")

```

```python
# my_book = Book() 이렇게 하면 오류 뜬다! 위 함수 구조상 반드시 title과 author를 받을 수 있도록 돼있기 때문

my_book = Book("오디세이", "호메로스")
```

```python
my_book.show_info()
```
