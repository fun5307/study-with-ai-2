### LCEL 연습하기
- prompt | model | output 형식

```python
# 일단 지금은 RAG와 상관 없음!

# 위의 from_templates와 달리 시스템 프롬프트 넣을 수 있음
prompt_test = ChatPromptTemplate.from_messages([
    ('system' , "너는 {character}야. 말투도 {character} 말투로 답변해"),   # 이런 식으로 넣을 내용 커스텀할 수 있다
    ('human' , "친절하게 질문에 답하도록 해. {question}")
])

chain = prompt_test | model | StrOutputParser()
result = chain.invoke({"character" : "어린아이", "question" : "오늘 밤에 할 일 추천해 줘!"})
print(result)
```

#### 출력형식 바꿔보기 - pydantic

```python
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field
from typing import List

class AnswerFormat(BaseModel):
    answer : str = Field(description="내일 할 일 추천 목록")
    recommend : List[str] = Field(description="내일 할 일 추천 목록")

parser = PydanticOutputParser(pydantic_object=AnswerFormat)

prompt_test = ChatPromptTemplate.from_messages([
    ('system' , "너는 {character}야. 말투도 {character} 말투로 답변해."
    "성격은 {feature}. 성격에 맞춰서 제대로 답을 해."
    "{format}에 맞춰서 답변하도록 해"),   # 이런 식으로 넣을 내용 커스텀할 수 있다
    ('human' , "친절하게 질문에 답하도록 해. {question}")
])

# 출력 형태 알아서 제안해주는 함수!
prompt = prompt_test.partial(
    format=parser.get_format_instructions())

prompt
```

```python
# 이거는 출력 형태 알아서 제안해주는 함수!
parser.get_format_instructions()
```

```python
chain = prompt | model | parser

result = chain.invoke({"character" : "어린아이", 
                       "question" : "오늘 밤에 할 일 추천해 줘!",
                       "feature" : "귀엽다"})
print(result)
```

```python
print("출력 1 - answer: ", result.answer)
print("출력 2 - recommend: ", result.recommend)
```

### chain 실습
- 주제를 정해서 프롬프트와, outputparser 바꿔서 체인으로 요청해보기
1. 프롬프트를 바꿔보세요
2. chain 연결해보세요
3. parser는 일단 stroutputparser로 고정

```python
prompt_prac = ChatPromptTemplate.from_messages([
    ('system' , "너는 {character}야. 질문에 대한 답변 언어는 {language}로 답변해"),   # 이런 식으로 넣을 내용 커스텀할 수 있다
    ('human' , "친절하게 질문에 답하도록 해. {question}")
])

chain = prompt_prac | model | StrOutputParser()
result = chain.invoke({"character" : "선생님",
                       "language" : "영어",
                         "question" : "Langchain에 대한 간략한 설명해줘"})
print(result)
```
