```python
from dotenv import load_dotenv
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings
from langchain_core.documents import Document

load_dotenv()

# 임베딩 및 저장
DB_PATH = "../data/k_ladder_2026"

# 임베딩 모델 설정
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
```

```python
# 저장된 벡터 DB 가져와야
load_vs = Chroma(
    collection_name="k_ladder_2026",
    embedding_function=embeddings,
    persist_directory=DB_PATH
)
```

### 검색기 설정
- 유사도로 찾습니다.
- mmr

```python
# 검색기
retriever = load_vs.as_retriever(
    search_kwargs = {"k" : 10}
)

# 실제로 검색
result = retriever.invoke("신혼 부부 지원 정책")
result
```

```python
# 검색기 2
retriever_mmr = load_vs.as_retriever(search_type="mmr", 
                                 search_kwargs={"k": 5, "fetch_k": 50, "lamda_mult" : 0.25})

# 실제로 검색
result_mmr = retriever_mmr.invoke("신혼 부부 지원 정책")
result_mmr
```

### LLM 호출 -> Langchain

```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage, HumanMessage

model = init_chat_model("openai:gpt-5.6-luna")

question = "대한민국의 청년 지원 정책 알려줘"
messages = [
    HumanMessage(content=question)
]

result = model.invoke(messages)
```

```python
print(result.content)
```

```python
# RAG로 붙여보기

SYSTEM_PROMPT = """
너는 공공정책 안내 도우미다.
아래 자료를 참고해서 답해라.
참고 자료에 없으면 "자료에 없음" 이라고 말해라.
정확한 자격, 금액, 기한은 공고 확인이 필요하다고 꼭 덧붙여라
답 끝에 참고한 페이지 번호를 [p.60]과 같은 형식으로 표시해라.
"""

# 검색 결과 문서를 받았을 때 메타데이터와 내용을 합쳐서 텍스트로 반환하는 함수 작성
def format_docs(docs):
    context = ""

    for doc in docs:
        context += f"[p.{doc.metadata["page"]}] \n {doc.page_content} \n\n"

    return context
```

```python
print(format_docs(result_mmr))
```

### 흐름
1. 질문을 한다
2. 문서를 찾는다
3. 문서를 합쳐서 컨텍스트에 넣는다
4. 최종 질문한다

```python
# 위 흐름을 함수로 만들어 보자! 이게 바로 RAG

def generate_answer(question):
    # 1. 질문에 해당하는 문서 찾기
    found_docs = retriever_mmr.invoke(question)

    # 2. 문서를 합쳐서 텍스트로 변환
    context = format_docs(found_docs)

    # 3. 최종 질문 프롬프트 작성
    prompt = f"""
            참고 자료
            {context}

            질문
            {question}
            """

    # 4. 메시지 작성
    messages = [
        SystemMessage(content=SYSTEM_PROMPT),
        HumanMessage(content=prompt)
    ]

    # 5. 최종 요청
    response = model.invoke(messages)

    return response, context
```

```python
result, context = generate_answer("구직 청년 지원 찾아줘")
print(result.content)
```
