### 반응형 차트
- 마우스 올리면 값이 뜨거나, 확대, 이동 가능한 차트 그리기
- 대시보드
- Streamlit -> 시각화 대시보드 만들기(나중에)

```python
import pandas as pd
import plotly.express as px     # 실제로 가장 많이 사용
import plotly.graph_objects as go # 그래프 생성의 세세한 부분까지 다 제어가능
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

## 아래 코드는 한글 사용할 수 있도록 해줌
plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/07-1_study_habits.csv") # 상대경로
df.head()
```

```python
fig = px.scatter(data_frame=df,
                 x='공부시간',
                 y='시험점수',
                 trendline="ols",
                 title="공부시간과 시험점수")

fig.show()
```

```python
# 코덱스가 만들어준 좀 더 이쁜(?) 그래프
fig = px.scatter(
      data_frame=df,
      x="공부시간",
      y="시험점수",
      trendline="ols",
      title="공부시간과 시험점수의 관계",
      labels={
          "공부시간": "공부 시간 (시간)",
          "시험점수": "시험 점수 (점)",
      },
      template="plotly_white",
      color_discrete_sequence=["#4C78A8"],
  )

fig.update_traces(
      marker=dict(size=10, opacity=0.75),
  )

fig.update_layout(
      title_x=0.5,
      title_font_size=22,
      xaxis=dict(showgrid=True, gridcolor="#EAEAEA"),
      yaxis=dict(showgrid=True, gridcolor="#EAEAEA"),
  )

fig.show()
```

```python
# 점 사이즈를 용돈으로 만들 수도 있다!
fig = px.scatter(data_frame=df,
                 x='공부시간',
                 y='시험점수',
                 color="반",
                #  trendline="ols",
                 size="용돈",
                 title="공부시간과 시험점수",
                 hover_data="성별")

fig.show()
```

### 다양한 차트 그리기
- 막대
- 원(pie): 비율

```python
mean = df.groupby("반", as_index=False)['시험점수'].mean()
print(mean)
```

```python
fig = px.bar(data_frame=mean, 
             x='반', 
             y='시험점수',
             color="반",
             title="월별 평균 점수")

fig.update_layout(width=600, height=400)
fig.show()
```

```python
# 파이차트
class_counts = df['반'].value_counts().reset_index()
class_counts
```

```python
fig = px.pie(data_frame=class_counts,
             names="반", 
             values="count",
             title="반별 인원 비율")
fig.update_layout(width=400, 
                  height=400, 
                  )  # 위쪽 여백)
fig.show()
```

```python
fig = px.pie(
      data_frame=class_counts,
      names="반",
      values="count",
      title="반별 인원 비율",
      hole=0.45,  # 가운데 빈 공간 비율
      color_discrete_sequence=["#4C78A8", "#F58518", "#54A24B", "#E45756"]
  )

# 조각 안에 반 이름과 퍼센트를 표시
fig.update_traces(
      textinfo="label+percent",
      textposition="inside",
      textfont_size=14,
      hovertemplate="%{label}<br>인원: %{value}명<br>비율: %{percent}<extra></extra>",
      pull=[0.03 if name != "C" else 0.15 for name in fig.data[0].labels]
)

fig.update_layout(
    width=500,
    height=500,
    title_x=0.5,  # 제목을 가운데 정렬
    margin=dict(t=70, b=70, l=30, r=30),
    legend=dict(orientation="h", y=-0.1, x=0.5, xanchor="center")
)

# 도넛 가운데에 전체 인원 표시
fig.add_annotation(
    text=f"전체<br>{class_counts['count'].sum()}명",
    x=0.5,
    y=0.5,
    showarrow=False,
    font=dict(size=18)
)

fig.show()
```

```python
# 히스토그램과 박스플롯

fig=px.box(data_frame=df,
           x='반',
           y='시험점수',
           color='반',
           points='all',
           width=600,
           height=500)

fig.show()
```

```python
# 히스토그램도 가능
fig = px.histogram(
      data_frame=df,
      x="시험점수",
      color="반",
      nbins=10,
      title="반별 시험점수 분포",
  )

fig.show()
```

### 여러 그래프를 한 화면에 나타내보기
- 여러 그래프를 한 화면에 모으면 그게 대시보드

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(rows=1, 
                    cols=2,
                    subplot_titles=["시험점수 분포", "공부시간 분포"])

fig.add_trace(go.Histogram(x=df['시험점수']), row=1, col=1) # go로 시작할 땐 그래프 이름 대문자!
fig.add_trace(go.Histogram(x=df['공부시간']), row=1, col=2)
fig.show()
```

```python
# 저장

fig.write_html("나의 첫 대시보드.html")
```
