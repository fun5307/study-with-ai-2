### 결과 해석, 의사 결정
- 통계 분석 목적 -> 더 나은 결정 내리려고
- 매출 건수로 매출 금액 계산하는 게 회귀

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from scipy import stats

plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/07-8_seoul_market_sales.csv")
df.head(3)
```

```python
r = stats.linregress(df['당월_매출_건수'], df['당월_매출_금액'])
r
```

```python
target_sales = 200_000_000
need_count = (target_sales - r.intercept) / r.slope
print(need_count)
```

```python
# 시각화

sns.regplot(data=df,
            x="당월_매출_건수",
            y="당월_매출_금액")

plt.show()
```

#### 다중 선형 회귀
- 여러 변수가 예측에 영향을 미칠 때
- 어떤 변수가 가장 중요한지 p값으로 가려볼 수 있다
- 숫자가 아닌 경우에는 계산 어렵다.
    - 범주형 변수 -> 숫자로 변환 필요

```python
data = pd.read_csv("../data/07-1_study_habits.csv")
data.head()
```

```python
import statsmodels.formula.api as smf

model = smf.ols("시험점수 ~공부시간 + 용돈", data=data).fit()
```

```python
print(model.rsquared)
```
