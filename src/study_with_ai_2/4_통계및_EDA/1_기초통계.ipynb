### 1. 데이터 불러오기

```python
import pandas as pd

df = pd.read_csv("../data/07-1_study_habits.csv") # 상대경로
# df = pd.read_csv(r"C:\study-with-ai-2\src\study_with_ai_2\data\07-1_study_habits.csv") # 절대경로

df.head()
```

### 2. 중심 - 평균, 중앙값, 최빈값

```python
mean_value = df["시험점수"].mean() # 평균
print(mean_value)

median_value = df["용돈"].median() # 중앙값
print(median_value)
```

```python
mode_value = df["시험점수"].mode()
print(mode_value) # 최빈값(영어로 mode) -> 타입이 series
print(mode_value.item(), mode_value[0])
```

### 3. 평균과 중앙값이 다른 경우

```python
print("용돈 평균:", df["용돈"].mean())
print("용돈 중앙값:", df["용돈"].median())
```

### 4. 범위, 분산, 표준편차
- 범위: 최댓값 ~ 최솟값
- 분산, 표준편차: 평균에서 얼마나 떨어져 있는지 보는 통계값

```python
scores = df['시험점수']
print("최솟값:", scores.max())
print("최대값:", scores.min())
print("범위:", scores.max()-scores.min())
```
