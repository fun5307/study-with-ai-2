### 1. 그룹 비교
- 반별로, 혹은 성별로 어떤 추세가 있는지, 집단별로 나눠서 분석해야 할 경우가 많다.
- 어느 그룹이 높고 낮은지 분석할 필요가 있다.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

## 아래 코드는 한글 사용할 수 있도록 해줌
plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/07-1_study_habits.csv") # 상대경로
df.head()
```

### 2. groupby로 그룹별 평균, 개수

```python
df.groupby("반")["시험점수"].mean().round(1)
```

### 3. 막대그래프 그리기

```python
sns.barplot(data=df, x='반', y='시험점수', hue='반')
plt.show()
```

```python
# 전체 차트 스타일 설정
sns.set_theme(style="whitegrid", palette="pastel")

plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="반",
    y="시험점수",
    hue="성별",          # 반마다 다른 색 사용
    legend=True       # 범례는 숨기기
)

plt.title("반별 평균 시험 점수", fontsize=16, fontweight="bold", pad=15)
plt.xlabel("반", fontsize=12)
plt.ylabel("평균 시험 점수", fontsize=12)

# plt.ylim(0, 100)       # 시험 점수가 100점 만점일 때
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()     # 글자가 잘리지 않게 조정
plt.show()
```

```python
# 반별 공부시간 평균을 막대그래프 그려보기

# mean_list = df.groupby("반")['공부시간'].mean().round(1)
plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False


sns.barplot(data=df, x='반', y='공부시간', hue='반', legend=True)
plt.title("반별 공부 시간 평균", fontsize=16, fontweight="bold", pad=15)
plt.xlabel("반", fontsize=12)
plt.ylabel("평균 공부 시간", fontsize=12)
plt.show()
```
