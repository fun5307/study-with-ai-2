- 관측 빈도와 기댓값을 비교해서 특징적인 부분을 판단

- 지금까지는 숫자를 비교
- 예) 성별에 다른 연령대 구성?

#### 1. 질문에 해당하는 내용 확정, 교차표 작성
- 성별에 따른 연령대 구성이 다른가?
- 성별과 연령대의 빈도를 교차표로 작성

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px  

plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/07-9_bike_ttareungi.csv")
df.head(3)
```

```python
data = df.dropna(subset="성별")
data.head(3)
```

```python
table = pd.crosstab(data['성별'], data['연령대코드'])
table
```

```python
from scipy import stats

chi2, p, dof, expected = stats.chi2_contingency(table)
print("카이제곱 통계량", chi2)
print("p-value", p)
print("자유도", dof)

# 결과: p가 0.05보다 낮으니 성별과 연령대 관련 있다!
```
