```python
import sqlite3

conn = sqlite3.connect("library.db")
cur = conn.cursor()
```

```python
cur.execute("""
CREATE TABLE books (
    book_id INTEGER PRIMARY KEY,
    title   TEXT,
    author  TEXT,
    year    INTEGER
    )
""")
```

### 데이터 넣기 - Insert

```python
book_list = [
    (1, "듄", "프랭크 허버트", 2021),
    (2, "은하수를 여행하는 히치하이커를 위한 안내서", "더글러스 애덤스", 1979),
    (3, "노르웨이의 숲", "무라카미 하루키", 1987),
    (4, "나미야 잡화점의 기적", "히가시노 게이고", 2012),
    (5, "더 기버", "로이스 로리", 1993)
]

cur.executemany("INSERT INTO books VALUES (?, ?, ?, ?)", book_list) # 물음표 4개인 이유: 카테고리가 4개라서
conn.commit()
```

```python
# 장르별 책 수를 합쳐 '책 수'라는 카테고리에 저장. genre도 '장르'라는 컬럼으로
# 문법: 일단 AS 뒤엔 따옴표 X, 그리고 함수들은 쉼표 안 쓰는듯

data = pd.read_sql("""
SELECT genre AS 장르,
COUNT(*) AS 책수    
FROM books_test
GROUP BY genre
""", conn)


data
```

```python
# 장르별 책 수를 합쳐 '책 수'라는 카테고리에 저장. genre도 '장르'라는 컬럼으로
# 쉼표 있고 없고 기준 : 한 함수(SELECT, FROM 등) 안에 여러 개 들어가면 쉼표로 구분, 함수와 함수 사이에는 쉼표 X

data = pd.read_sql("""
SELECT genre AS 장르,
    COUNT(*) AS 책수,
    AVG(PRICE) AS 평균가격
FROM books_test
GROUP BY genre
ORDER BY 책수
""", conn)

data
```
